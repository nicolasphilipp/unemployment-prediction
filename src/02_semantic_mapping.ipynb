# WP2 – Semantic Mapping

## Task T2.2 – Semantic Mapping

This documents the semantic mapping of all database attributes to ontology concepts for the FAIR Data Science project:

**Prediction of unemployment in Vienna districts using tourism and demographic data**

The semantic mappings were created to improve the FAIRness of the data, especially:

- **Findability (F)** through standardized semantic annotations
- **Interoperability (I)** through ontology-based metadata

The mappings are intended for integration into DBRepo metadata via the REST API.

## Ontologies Used

The semantic mappings mainly use concepts from the following ontologies:

### 1. SDMX (Statistical Data and Metadata eXchange)
URI base: https://sdmx.org/

SDMX is a widely used international standard for statistical datasets and official government statistics. Since the experiment uses regional unemployment and tourism statistics published by the City of Vienna, SDMX concepts are suitable for describing statistical regions, reference periods, and measured indicators.

### 2. QUDT (Quantities, Units, Dimensions and Data Types)
URI base: http://qudt.org/

QUDT is a well-established ontology for describing measurable quantities and units. It is used to describe densities, counts, population values, and measurement-related numerical attributes.

### 3. GeoNames Ontology
URI base: https://www.geonames.org/ontology

GeoNames provides semantic concepts for geographic and administrative regions. It is appropriate for district-level regional identifiers and location-related metadata.

### 4. Time Ontology in OWL
URI base: https://www.w3.org/TR/owl-time/

The W3C Time Ontology provides semantic concepts for dates and temporal references. It is suitable for describing reference years and dates used in time-series statistical data.
