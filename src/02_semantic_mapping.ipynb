## Task T2.2 – Semantic Mapping

This documents the semantic mapping of all database attributes to ontology concepts for the FAIR Data Science project:

**Prediction of unemployment in Vienna districts using tourism and demographic data**

The semantic mappings were created to improve the FAIRness of the data, especially:

- **Findability (F)** through standardized semantic annotations
- **Interoperability (I)** through ontology-based metadata

The mappings are intended for integration into DBRepo metadata via the REST API.

## Ontologies Used

The semantic mappings mainly use concepts from the following ontologies:

### 1. SDMX (Statistical Data and Metadata eXchange)
URI base: http://purl.org/linked-data/sdmx/2009/

SDMX is a widely used international standard for statistical datasets and official government statistics. Since this experiment uses regional unemployment and tourism statistics published by the City of Vienna, specific SDMX concepts are used to describe observation values and demographic dimensions. This standard ensures high domain-specific interoperability and matches the structural formatting used by public administration reporting systems.

### 2. QUDT (Quantities, Units, Dimensions and Data Types)
URI base: http://qudt.org/

QUDT is a well-established ontology for describing measurable quantities, units, and data dimensions. It is utilized here to explicitly declare the unit metrics for densities, counts, and measurement-related numerical attributes. Applying QUDT concepts prevents the misinterpretation of numerical scales and statistical proportions when external researchers interact with or reuse the data model.

### 3. GeoNames Ontology
URI base: https://www.geonames.org/ontology

GeoNames provides semantic concepts for geographic, spatial, and administrative regions. It is chosen for this project to map district-level regional identifiers and specific administrative division codes to a standardized global vocabulary. This mapping removes regional naming ambiguities and ensures that local Vienna district data can be instantly integrated into external geographic information systems (GIS).

### 4. Time Ontology in OWL
URI base: https://www.w3.org/TR/owl-time/

The W3C Time Ontology provides semantic concepts for dates, durations, and temporal references. It is suitable for describing reference years and observation dates used across our time-series statistical tables. Using this ontology explicitly defines temporal bounds as machine-readable entities, which allows automated semantic scrapers to correctly build chronological data timelines.
